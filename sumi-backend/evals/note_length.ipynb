```python
from pathlib import Path

import matplotlib.pyplot as plt
```

```python
data_dir = "../../data/notion-export-markdown/"
```

```python
def get_filepaths(root_dir: str, allowed_file_extensions: list[str] = None) -> list[str]:
    paths = Path(root_dir).rglob("*")
    return [
        str(p) for p in paths
        if p.is_file() and (allowed_file_extensions is None or p.suffix in allowed_file_extensions)
    ]
```

```python
filepaths = get_filepaths(data_dir, [".md"])
len(filepaths)
```

```python
character_length_distribution = []
for file in filepaths:
    with open(file, "r", encoding="utf-8") as f:
        content = f.read().strip()
        character_length_distribution.append(len(content))
```

```python
plt.hist(character_length_distribution, bins=30)
```

```python
for idx, length in enumerate(character_length_distribution):
    if length > 1e5:
        print(idx)
```
